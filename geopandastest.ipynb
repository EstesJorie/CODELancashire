```python
southPath = get_path('geoda south')
south = gp.read_file(southPath)
```

```python
dataPath = get_path('nybb')
nybb = gp.read_file(dataPath)
```

```python
earthData = get_path('naturalearthland')
earth = gp.read_file(earthData)
```

```python
south.head()
south.keys()

south['HR*'] =  (south['HR60'] + south['HR70'] + south['HR80'] + south['HR90']).astype(int)
south['HR*'] = statistics.mean(south['HR*'])
```
